```python
# ----- 1. Percentile-Based Demographic Parity (DP) -----
# Define percentiles corresponding to top 20% (80th percentile) and top 30% (70th percentile)
percentile_thresholds = {"top_20%": 80, "top_30%": 70}
percentile_results = {}
for label, pct in percentile_thresholds.items():
    threshold = np.percentile(read_df['SMOG'], pct)
    # Identify high-complexity cases
    read_df['high_complexity'] = (read_df['SMOG'] >= threshold).astype(int)
    # Compute selection rates by group
    rates = read_df.groupby('Class')['high_complexity'].mean()
    # Calculate difference: privileged minus underprivileged
    diff = rates.get('Privileged', np.nan) - rates.get('Underprivileged', np.nan)
    percentile_results[label] = {
        'threshold': threshold,
        'privileged_rate': rates.get('Privileged', np.nan),
        'underprivileged_rate': rates.get('Underprivileged', np.nan),
        'difference': diff
    }
# Display the results
import pandas as pd
pd.DataFrame(percentile_results).T
```

```python
# ----- 2. Distributional Demographic Parity (DP) ----
# Separate scores by group
priv_scores = read_df.loc[read_df['Class'] == 'Privileged', 'SMOG'].values
unpriv_scores = read_df.loc[read_df['Class'] == 'Underprivileged', 'SMOG'].values
# Wasserstein distance measures the difference between two distributions (location, spread, mass)
wasserstein_dist = wasserstein_distance(priv_scores, unpriv_scores)
# KL divergence requires discrete probability distributions; use histogram bins
num_bins = 'auto'
bins = np.histogram_bin_edges(read_df['SMOG'], bins=num_bins)
priv_hist, _ = np.histogram(priv_scores, bins=bins, density=True)
unpriv_hist, _ = np.histogram(unpriv_scores, bins=bins, density=True)
# Avoid zero entries to prevent division by zero in KL computation
epsilon = 1e-10
priv_hist += epsilon
unpriv_hist += epsilon
# Normalize histograms to sum to 1
priv_hist /= priv_hist.sum()
unpriv_hist /= unpriv_hist.sum()
# KL divergence is asymmetric; compute both directions
kl_priv_to_unpriv = entropy(priv_hist, unpriv_hist)
kl_unpriv_to_priv = entropy(unpriv_hist, priv_hist)
# Summarize results
{
    'Wasserstein_distance': wasserstein_dist,
    'KL_priv_to_unpriv': kl_priv_to_unpriv,
    'KL_unpriv_to_priv': kl_unpriv_to_priv
}
```

```python
# ----- 3. Counterfactual Demographic Parity (DP) -----
def counterfactual_dp_metric(df, metric='SMOG'):
    count = 0
    total_pairs = 0
    # Group by note_id and model to compare identical reports
    grouped = df.groupby(['note_id', 'model'])
    for (_, _), group in grouped:
        priv_scores = group.loc[group['Class'] == 'Privileged', metric].values
        unpriv_scores = group.loc[group['Class'] == 'Underprivileged', metric].values
        if len(priv_scores) == 0 or len(unpriv_scores) == 0:
            continue
        # Compare all pairs of privileged and underprivileged scores
        # Count instances where privileged score exceeds underprivileged score
        for ps in priv_scores:
            count += np.sum(ps > unpriv_scores)
        total_pairs += len(priv_scores) * len(unpriv_scores)
    return count / total_pairs if total_pairs > 0 else np.nan
# Compute counterfactual DP on SMOG
counterfactual_dp_score = counterfactual_dp_metric(read_df, metric='SMOG')
counterfactual_dp_score
```
